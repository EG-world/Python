### 1-2. cursor 생성하기
* 하나의 데이터베이스 connection에 대해 독립적으로 SQL문을 실행할 수 있는 작업환경 제공하는 객체

### 1-3. SQL문의 결과 가져오기
* fetchall() : 한번에 모든 tuple을 가져옴. 검색 결과가 매우 많다면 메모리 오버헤드(넘쳐 오류가 생기는 현상)가 발생할 수 있음.
* fetchone() : 한번에 하나의 tuple을 가져옴. 다시 fetchone() 메서드를 호출하면 다음 데이터를 사져옴

````python
data = cur.fetchall() # 저장한 값을 출력
print(data)
````

````python
data = cur.fetchall() # flash라는 구조로 한번 보여주면 띄워놓은 데이터를 지움.
print(data)
````

### 문제2
* 회원가입 프로그램을 만들어보자
* 단, 중복 데이터로 인한 오류시 "다시 입력하세요" 라는 오류처리와 함께 다시 등록할 수 있도록 함
* 단, 회원가입 후 "추가로 가입하시겠습니까? (Y/N)" 를 입력받아 추가로 입력할 수 있도록 함

```
아이디를 입력하세요 : xxx
비밀번호를 입력하세요 : xxxx
이름을 입력하세요 : xxx
전화번호를 입력하세요 : xxx
이메일을 입력하세요 : xxx
성별을 입력하세요 : xxx
주민번호 앞자리를 입력하세요 : xxx
주민번호 뒷자리를 입력하세요 : xxx
우편번호를 입력하세요 : xxx
상세주소를 입력하세요 : xxxx
참고사항을 입력하세요 : xxx
```

````python
import MySQLdb
````

````python
db = MySQLdb.connect(host='localhost', user='root', password='1234', db='kdt')
cur = db.cursor()

class Sign :
    def start(self) :
        filtering = input('회원가입 진행하시겠습니까? (Y/N) : ').strip().upper()
        if filtering == 'Y' :
            while True:
                data = []
                try:
                    id = input('아이디를 입력하세요 : ')
                    pw = input('비밀번호를 입력하세요 : ')
                    name = input('이름을 입력하세요 : ')
                    hp = input('전화번호를 입력하세요 : ')
                    email = input('이메일을 입력하세요 : ')
                    gender = input('성별을 입력하세요 : ')
                    ssn1 = input('주민번호 앞자리를 입력하세요 : ')
                    ssn2 = input('주민번호 뒷자리를 입력하세요 : ')
                    zipcode = input('우편번호를 입력하세요 : ')
                    address1 = input('상세주소를 입력하세요 : ')
                    address2 = input('참고사항을 입력하세요 : ')
                    sql = 'insert into member(userid, userpw, name, hp, email, gender, ssn1, ssn2, zipcode, address1, address2 values(%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)'
                    data.append((id, pw, name, email, gender, ssn1, ssn2, zipcode, address1, address2))
                    cur.executemany(sql,data)
                    db.commit()
                    ask = input('추가로 가입하시겠습니까? (Y/N) : ').strip().upper()
                    if ask == 'Y' :
                        continue
                    elif ask == 'N' :
                        break
                    else :
                        print('잘못입력하셨습니다 초기화면으로 돌아갑니다.')

                except :
                    print('다시 입력하세요')
        else :
            print('종료합니다')
````

````python
user = Sign()
````

````python
user.start()
cur.close()
db.close()
````

### 문제3
* 로그인 프로그램을 작성해보자
* 단, 아이디와 비밀번호가 맞은 유저가 있을 경우 "로그인 되었습니다", 유저가 없을 경우 "아이디 또는 비밀번호를 확인하세요" 라고 출력
```
아이디를 입력하세요 : xxx
비밀번호를 입력하세요 : xxx
```

````python
class Login :
    def login(self) :
            db = MySQLdb.connect(host='localhost', user='root', password='1234', db='kdt')
            cur = db.cursor(MySQLdb.cursors.DictCursor)
            sql = 'select userid, userpw from member'
            cur.execute(sql)
            data = cur.fetchall()
            for row in data :
                id = input('아이디를 입력하세요 : ')
                if row['userid'] == id :
                    pw = input('비밀번호를 입력하세요 : ')
                    if row['userpw'] == pw :
                        print('로그인 되었습니다')
                        break
                    else :
                        print('비밀번호가 틀렸습니다.')
                        break
                else :
                    print("존재하지 않는 유저입니다.")
                    break
            cur.close()
            db.close()
````
